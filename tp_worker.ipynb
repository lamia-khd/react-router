```python
import time # Imports the module used to measure elapsed time
from concurrent.futures import ThreadPoolExecutor # Imports the tool needed to manage parallel execution

TEMPS = {"Blida": 21, "Alger": 22, "Oran": 23, "Annaba": 20, "Setif": 15, "Bechar": 30} # Provided dictionary storing the temperatures
def get_weather(city): # Defines the function that queries our simulated weather server
    """Weather API: today's temperature of one city (1 second).""" # Documentation string explaining the function
    time.sleep(1) # Simulates the mandatory one-second waiting time from the server
    return TEMPS[city] # Returns the specific temperature for the requested city
cities = list(TEMPS) # Creates a list of the 6 cities extracted from the dictionary keys

team_sizes = [1, 2, 3, 6, 12] # Defines the different team sizes we need to test for Q2 and Q4
for workers in team_sizes: # Loops through each team size to avoid repeating the same code
    t = time.perf_counter() # Starts the stopwatch just before launching the queries
    with ThreadPoolExecutor(max_workers=workers) as team: # Creates a pool with the specified number of workers
        results = list(team.map(get_weather, cities)) # Assigns the get_weather task for each city to the team
    Tp = time.perf_counter() - t # Stops the stopwatch and calculates the total duration

    question = "Q2" if workers <= 6 else "Q4" # Assigns the Q2 or Q4 label based on the current worker count
    print(f"{question} {workers} workers: {Tp:.1f} s") # Prints the formatted execution time for the current team
```

### Answers to the Questions

* **Q1 (Guess):** 1 worker = **6s**, 2 workers = **3s**, 3 workers = **2s**, 6 workers = **1s**.
* **Q2 (Measured):** 1 worker = **6.0s**, 2 workers = **3.0s**, 3 workers = **2.0s**, 6 workers = **1.0s**.
* **Q3 (Why 2 workers = 3s?):** The 6 tasks are processed in 3 sequential batches of 2 parallel tasks each (3 batches × 1s = 3s).
* **Q4 (12 workers?):** **1.0s**. We only have 6 tasks, so 6 workers do the work in parallel in 1s while the other 6 remain idle.
